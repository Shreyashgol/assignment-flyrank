**Lane:** Refresh / Content Opportunity Scoring

**Why:** It directly addresses the practical need of editors deciding what to work on next. It provides actionable outputs based on safe, observable signals rather than trying to guess a black-box algorithm. Ranking candidate pages helps allocate limited human editing time efficiently.

```python
import pandas as pd
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')
# Confirm we have a large queue of pages that need review
print(f"Total pages to potentially review: {len(df):,}")
```

## 2. The question: decision, action, cost of a wrong call

**Decision:** Which visible pages should a content editor review and update first?
**Action:** A content editor rewrites the title/meta description, expands thin content, or updates stale information to regain visibility.
**Cost of a wrong call:** Wasted editor hours working on a page that doesn't need it or won't benefit, while missing out on rescuing a high-value page that is actively decaying.

```python
# Example of a costly wrong call: a page with high impressions but low age that isn't actually declining yet.
costly_mistake_candidates = df[(df['impressions_90d'] > 1000) & (df['content_age_days'] < 90)]
print(f"Pages that are too new to safely touch, but have high impressions: {len(costly_mistake_candidates)}")
print("An editor wasting time here is a missed opportunity elsewhere.")
```

## 3. Quick look at the data (2-3 real numbers)

Looking at the starter dataset, we have 30,000 pages that generated ~156 million impressions in the last 90 days. Strikingly, over half of these pages (16,262) are trending downwards. This massive volume of declining content means a simple human review is impossible, and an intelligent ranking system is strictly necessary to prioritize rescues.

```python
import pandas as pd
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')
print(f"Total pages: {len(df)}")
print(f"Pages trending down: {len(df[df['trend_direction'] == 'down'])}")
print(f"Total 90d impressions: {df['impressions_90d'].sum():,}")
```

## 4. Careful words: what I can and can't claim

**What I can claim:** We can use this to provide *decision-support* to rank review candidates based on *observed* search and engagement signals. We can show *directional* evidence that certain pages are losing visibility compared to their baseline.
**What I cannot claim:** We cannot claim we are *predicting Google's algorithm*, nor can we provide *causal proof* that a refresh will guarantee a recovery without running a controlled experiment.

```python
# Show that we are using strictly observed metrics (impressions, sessions, etc.)
observed_cols = ['impressions_90d', 'sessions_90d', 'avg_position', 'trend_direction']
print("We build our ranking strictly on observed past signals:")
print(df[observed_cols].head(3))
print("\nNotice there are no 'Google algorithm' flags here—just recorded traffic and position.")
```
